```python
import pandas as pd

# 1. Define your file names
input_file = "student_performance_dataset_bigger.csv"
output_file = "student_performance_sample_50k.csv"

print("Loading original dataset...")
df = pd.read_csv(input_file)

# 2. Randomly sample 50,000 rows
# We use random_state=42 so that if you run this twice, you get the exact same random mix
print("Sampling 50,000 rows...")
df_sampled = df.sample(n=50000, random_state=42)

# 3. Save the new, smaller dataset
print(f"Saving sample to {output_file}...")
df_sampled.to_csv(output_file, index=False)

print("Done! You can now upload the new file.")
```

```python
import pandas as pd
import numpy as np

# Set the file names
INPUT_FILE = "student_performance_dataset_bigger.csv"
OUTPUT_FILE = "student_performance_augmented.csv"

print("Loading original dataset. This might take a moment for 1M rows...")
df = pd.read_csv(INPUT_FILE)

print("Engineering new Academic History features...")
np.random.seed(42) # Ensures reproducibility 

# 1. Generate Continuous Assessment Score (Out of 40)
# We base this heavily on the final exam score, but add random real-world variance.
# Assuming final exam score (which is out of 100) translates roughly to CA performance.
base_ca = (df['exam_score'] / 100) * 40

# Add realistic variance (standard deviation of 3 points)
noise_ca = np.random.normal(loc=0, scale=3.0, size=len(df))
df['continuous_assessment_score'] = np.clip(base_ca + noise_ca, 0, 40).round(1)

# 2. Generate Previous Semester GPA (Out of 4.0)
# Students who score well generally have higher past GPAs.
# 100 roughly maps to 4.0, 50 roughly maps to 2.0.
base_gpa = (df['exam_score'] / 100) * 4.0

# Add realistic variance (standard deviation of 0.4 GPA points)
# We also boost it slightly because people usually have higher GPAs than single exam scores
noise_gpa = np.random.normal(loc=0.2, scale=0.4, size=len(df))
df['previous_semester_gpa'] = np.clip(base_gpa + noise_gpa, 0.0, 4.0).round(2)

# Optional: Tweak behavioural correlation slightly to simulate high-intent studying
# (Students with high GPA and high study hours get a slight bump in realism)
print("Applying final formatting...")

# Reorder columns so the new ones sit right before the exam_score
cols = df.columns.tolist()
# Remove the ones we just added from the end
cols.remove('continuous_assessment_score')
cols.remove('previous_semester_gpa')

# Find the index of exam_score to insert the new features right before it
exam_idx = cols.index('exam_score')
new_col_order = (
    cols[:exam_idx] + 
    ['continuous_assessment_score', 'previous_semester_gpa'] + 
    cols[exam_idx:]
)
df = df[new_col_order]

print(f"Saving augmented dataset to '{OUTPUT_FILE}'...")
df.to_csv(OUTPUT_FILE, index=False)

print("Done! Here is a preview of your new dataset:")
print(df[['study_hours_per_day', 'continuous_assessment_score', 'previous_semester_gpa', 'exam_score']].head())
```
